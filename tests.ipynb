```python
import os
import requests
import dotenv
import pandas as pd
import datetime
```

```python
dotenv.load_dotenv('.env.prod')
import dotenv

api_key = dotenv.get_key('.env.prod', 'ALPHAVANTAGE_API_KEY')
symbol = 'AAPL'
url = f"https://www.alphavantage.co/query?function=HISTORICAL_OPTIONS&symbol={symbol}&require_greeks=true&apikey={api_key}"
response = requests.get(url)
data = response.json()
print(data)
```

```python
data = data.get("data", [])
data_df = pd.DataFrame(data)
data_df
```

```python
data_df.to_csv('data.csv') 
```

```python
target_delta_range = (0.5, 1.0)
target_dtm_range = (15, 30)
today = datetime.datetime.now()
time_range_low = today - datetime.timedelta(days=target_dtm_range[0])
time_range_high = today + datetime.timedelta(days=target_dtm_range[1])
data_df['expiration'] = pd.to_datetime(data_df['expiration'])
data_df['delta'] = data_df['delta'].astype(float)
mask = (
    (data_df['delta'].abs() >= target_delta_range[0])
    & (data_df['delta'].abs() <= target_delta_range[1])

)
filtered_data = data_df[mask] 
filtered_data
```

```python
filtered_data.to_csv('filtered_data.csv')
```
